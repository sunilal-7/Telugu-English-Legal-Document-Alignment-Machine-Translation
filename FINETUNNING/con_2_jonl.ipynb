```python
import pandas as pd
import json

# Load CSV (use your actual CSV path if needed)
df = pd.read_csv("aligned_telugu_eng_output.csv")

# Rename columns
df = df.rename(columns={
    "English_Reference": "eng_Latn",
    "Telugu": "tel_Telu"
})

# Keep required columns
df = df[["eng_Latn", "tel_Telu"]]

# Drop empty rows
df = df.dropna()

# Save JSONL in current directory
output_path = "eng_tel_.jsonl"

with open(output_path, "w", encoding="utf-8") as f:
    for _, row in df.iterrows():
        record = {
            "eng_Latn": str(row["eng_Latn"]).strip(),
            "tel_Telu": str(row["tel_Telu"]).strip()
        }
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

print(f"✅ Saved {len(df)} lines to {output_path}")
```
